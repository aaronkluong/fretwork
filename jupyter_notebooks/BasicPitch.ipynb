```python
model_output, midi_data, note_events = predict(audio_filename)

print(f"Got {len(note_events)} notes\n")
print("First 10 notes:")
for note in note_events[:10]:
    start, end, pitch_midi, amplitude, pitch_bends = note
    print(f"  pitch={pitch_midi} (MIDI), start={start:.2f}s, end={end:.2f}s")
```

```python
def midi_to_note_name(midi_number):
    notes = ['C', 'C#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'B']
    octave = (midi_number // 12) - 1
    note = notes[midi_number % 12]
    return f"{note}{octave}"

# Sort chronologically and show first 20
note_events_sorted = sorted(note_events, key=lambda n: n[0])
print("First 20 notes (chronological):")
for note in note_events_sorted[:20]:
    start, end, pitch_midi, amplitude, _ = note
    note_name = midi_to_note_name(pitch_midi)
    duration = end - start
    print(f"  {note_name:4s}  start={start:5.2f}s  dur={duration:.2f}s  amp={amplitude:.2f}")
```
