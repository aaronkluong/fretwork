```python
def normalize_chord_label(label):
    """Convert autochord's flat naming to guitarist-friendly sharp naming."""
    enharmonic_map = {
        'Ab': 'G#',
        'Bb': 'A#',
        'Db': 'C#',
        'Eb': 'D#',
        'Gb': 'F#',
    }
    if label == 'N':
        return label
    for flat, sharp in enharmonic_map.items():
        if label.startswith(flat):
            return label.replace(flat, sharp, 1)
    return label

```

```python
chords = autochord.recognize(audio_filename, lab_fn='chords.lab')

print("Normalized chord progression:")
for start, end, label in chords:
    duration = end - start
    norm_label = normalize_chord_label(label)
    print(f"  {start:5.2f}s → {end:5.2f}s  ({duration:.2f}s)  {norm_label}")
```
